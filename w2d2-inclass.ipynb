## Decorators


````python
import functools
import time

def timer_decorator(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        print(time.perf_counter() - start, 's elapsed')
        return result
    return wrapper

@timer_decorator
def square(n):
    time.sleep(1.0)
    return n * n

N = 2**8

print(square(N))

print('name of fn', square.__name__)
````

````python
def authenticate(fixed_api_key):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            return func(*args, api_key=fixed_api_key, **kwargs)  # Call og function with key
        return wrapper
    return decorator

# wrapped_function = decorator_with_arg(arg)(original function)

# -- Dummy API interface functions --------------------------------

@authenticate(user_1_key)
def api_get(endpoint, api_key):
    data = f"Got data from {endpoint} using key {api_key}"
    print(data)
    return data

@authenticate(user_1_key)
def api_post(endpoint, payload, api_key):
    response = f"Posted {payload} to {endpoint} using key {api_key}"
    print(response)
    return response

@authenticate(user_1_key)
def api_patch(endpoint, payload, api_key):
    response = f"Patched {payload} to {endpoint} using key {api_key}"
    print(response)
    return response

@authenticate(user_1_key)
def api_delete(endpoint, api_key):
    response = f"Deleted at {endpoint} using key {api_key}"
    print(response)
    return response

# -----------------------------------------------------------------

data = api_get("/data/1")
resp = api_post("/data/1", {"name": "Alice"})
resp = api_patch("/data/1", {"name": "Alice Smith"})
resp = api_delete("/data/1")

# ---

print('__name__ of api_get:', api_get.__name__)
print('__name__ of api_post:', api_post.__name__)
print('__name__ of api_patch:', api_patch.__name__)
print('__name__ of api_delete:', api_delete.__name__)
````

````python
log_msgs = []  # Logging collection

def authenticate(fixed_api_key):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            return func(*args, api_key=fixed_api_key, **kwargs)  # Call og function with key
        return wrapper
    return decorator

def log(log_msgs: list):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            timestamp = time.strftime("%Y-%m-%d %H:%M:%S", time.localtime())
            endpoint = args[0]
            op = func.__name__
            log_entry = timestamp + ' ' + endpoint + ' ' + op
            log_msgs.append(log_entry)
            return func(*args, **kwargs)
        return wrapper
    return decorator

# -- Dummy API interface functions --------------------------------

@authenticate(user_1_key)
@log(log_msgs)
def api_get(endpoint, api_key):
    data = f"Got data from {endpoint} using key {api_key}"
    print(data)
    return data

@authenticate(user_1_key)
@log(log_msgs)
def api_post(endpoint, payload, api_key):
    response = f"Posted {payload} to {endpoint} using key {api_key}"
    print(response)
    return response

@authenticate(user_1_key)
@log(log_msgs)
def api_patch(endpoint, payload, api_key):
    response = f"Patched {payload} to {endpoint} using key {api_key}"
    print(response)
    return response

@authenticate(user_1_key)
@log(log_msgs)
def api_delete(endpoint, api_key):
    response = f"Deleted at {endpoint} using key {api_key}"
    print(response)
    return response

# -----------------------------------------------------------------

data = api_get("/data/1")
data = api_get("/data/2")
data = api_get("/data/3")
resp = api_post("/data/1", {"name": "Alice"})
resp = api_patch("/data/1", {"name": "Alice Smith"})
resp = api_delete("/data/1")

# ---

print("\nLog\n-----\n" + '\n'.join(log_msgs))
````

## Generators
Allow us to add logic to an iterator *that is lazily executed* with the `yield` keyword

````python
def countdown(n):
    i = n
    while i > 0:
        time.sleep(1.0)
        yield i
        i -= 1

# --------------------------------------------------------

for i in countdown(5):
    print(i)

print("Done!")
````

````python
# Fibonacci sequence
# The Fibonacci sequence is an infinite sequence
# of numbers where each number is the sum of the
# two preceding ones, usually starting with 0 and 1.

N = 10

nfib = [0, 1]

for i in range(2, N):
    nfib.append(nfib[i - 1] + nfib[i - 2])
    print('nfib now contains', len(nfib))

for num in nfib:
    print(num)

print(type(nfib))
````

````python
def fibgen(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b

nfib = fibgen(10)

for num in nfib:  # Where the numbers get generated
    print(num)

print(nfib[5])

print(type(nfib))
````

Generators can be chained together...

````python
squares = [x**2 for x in range(10)]
squares_gen = (x**2 for x in range(10))

print(squares[5])
print(squares_gen[5])
````

````python
import random

def even(gen):
    for num in gen:
        if num % 2 == 0:
            yield num

def square(gen):
    for num in gen:
        yield num ** 2

def inf_randint(lower, upper):
    while True:  # Remember to break!
        yield random.randint(lower, upper)

for num in (
    square(even(inf_randint(1, 10)))
):
    print(num)
    if num > 9:
        break
````
